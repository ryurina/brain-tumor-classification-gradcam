```python
import torch
import torch.nn as nn
import torch.optim as optim

import numpy as np

from torchvision import datasets, transforms
from torchvision.models import resnet18, ResNet18_Weights

from torch.utils.data import DataLoader, Subset
from sklearn.model_selection import train_test_split

from tqdm import tqdm
import copy

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)
```

### 1. Pretraîtement des données

```python

# RESNET18 normalisation
mean = [0.485, 0.456, 0.406]
std = [0.229, 0.224, 0.225]

train_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.RandomRotation(10),
    transforms.ToTensor(),
    transforms.Normalize(mean, std)
])

val_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean, std)
])

test_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean, std)
])

# Load the same training directory with different transforms
train_aug = datasets.ImageFolder(
    "../dataset/Training",
    transform=train_transform
)

full_train_eval = datasets.ImageFolder(
    "../dataset/Training",
    transform=val_transform
)

test_dataset = datasets.ImageFolder(
    "../dataset/Testing",
    transform=test_transform
)

# avoir les  label
targets = train_aug.targets
indices = list(range(len(targets)))

# separer en deux groupes train/val
train_idx, val_idx = train_test_split(
    indices,
    test_size=0.2,
    stratify=targets,
    random_state=42
)


train_dataset = Subset(train_aug, train_idx)
val_dataset = Subset(full_train_eval, val_idx)


train_loader = DataLoader(
    train_dataset, batch_size=32,
    shuffle=True, num_workers=4
)

val_loader = DataLoader(
    val_dataset, batch_size=32,
    shuffle=False, num_workers=4
)


test_loader = DataLoader(
    test_dataset,
    batch_size=32,
    shuffle=False,
    num_workers=4,
    pin_memory=True
)

print("Training:", len(train_dataset))
print("Validation:", len(val_dataset))
print("Test:", len(test_loader))
```

### 2. ResNET18

```python
# Hyperparametres

NUM_CLASSES = 4
LEARNING_RATE = 1e-4

images, labels = next(iter(train_loader))

print("Images:", images.shape)
print("Labels:", labels.shape)
print("Classes:", train_aug.classes)
```

```python
model = resnet18(
    weights=ResNet18_Weights.DEFAULT
)

model.fc = nn.Linear(
    model.fc.in_features,
    NUM_CLASSES
)

model = model.to(device)

criterion = nn.CrossEntropyLoss()

optimizer = optim.Adam(
    model.parameters(),
    lr=LEARNING_RATE
)

print("ResNet18 prêt!")
```

```python
def train_one_epoch(model, loader, criterion, optimizer, device):

    model.train()

    running_loss = 0.0
    correct = 0
    total = 0

    for images, labels in tqdm(loader, leave=False):

        images = images.to(device)
        labels = labels.to(device)
        
        optimizer.zero_grad()
    
        outputs = model(images)

        loss = criterion(outputs, labels)
        
        loss.backward()

        optimizer.step()

        running_loss += loss.item() * images.size(0)

        _, predicted = torch.max(outputs, 1)

        total += labels.size(0)
        correct += (predicted == labels).sum().item()

    epoch_loss = running_loss / total
    epoch_acc = 100 * correct / total

    return epoch_loss, epoch_acc
```

```python
def validate(model, loader, criterion, device):

    model.eval()

    running_loss = 0.0
    correct = 0
    total = 0

    with torch.no_grad():

        for images, labels in loader:

            images = images.to(device)
            labels = labels.to(device)

            outputs = model(images)

            loss = criterion(outputs, labels)

            running_loss += loss.item() * images.size(0)

            _, predicted = torch.max(outputs, 1)

            total += labels.size(0)
            correct += (predicted == labels).sum().item()

    epoch_loss = running_loss / total
    epoch_acc = 100 * correct / total

    return epoch_loss, epoch_acc
```

### 3. Entrainement

```python
best_val_acc = 0.0

history = {
    "train_loss": [],
    "train_acc": [],
    "val_loss": [],
    "val_acc": []
}

for epoch in range(EPOCHS):

    print(f"\nEpoch [{epoch+1}/{EPOCHS}]")

    # Training
    train_loss, train_acc = train_one_epoch(
        model, train_loader, criterion, optimizer, device
    )

    # Validation
    val_loss, val_acc = validate(
        model, val_loader, criterion, device
    )

    # historique
    history["train_loss"].append(train_loss)
    history["train_acc"].append(train_acc)
    history["val_loss"].append(val_loss)
    history["val_acc"].append(val_acc)

    print(f"Train Loss: {train_loss:.4f} | Train Acc: {train_acc:.2f}%")
    print(f"Val Loss:   {val_loss:.4f} | Val Acc:   {val_acc:.2f}%")

    if val_acc > best_val_acc:

        best_val_acc = val_acc

        torch.save(
            model.state_dict(),
            "../models/best_resnet18.pth"
        )

        print("Enregistrement du meilleur model")

print("Entraînement complete!")
print(f"Validation Accuracy: {best_val_acc:.2f}%")
```

### 4. Evaluation du model

```python
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    accuracy_score
)

all_preds = []
all_labels = []

with torch.no_grad():
    for images, labels in test_loader:

        images = images.to(device)

        outputs = model(images)
        predictions = outputs.argmax(dim=1)

        all_preds.extend(predictions.cpu().numpy())
        all_labels.extend(labels.numpy())

# accuracy
test_acc = accuracy_score(all_labels, all_preds)

print(f"Test Accuracy: {test_acc * 100:.2f}%")

print(classification_report(
    all_labels,
    all_preds,
    target_names=train_aug.classes,
    digits=4
))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

cm = confusion_matrix(all_labels, all_preds)

plt.figure(figsize=(8, 6))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=train_aug.classes,
    yticklabels=train_aug.classes
)

plt.xlabel("Prediction Label")
plt.ylabel("Correct Label")
plt.title("ResNet18 - Test Matrice de Confusion")

plt.tight_layout()
plt.show()
```

### 5. GradCAM

```python
import torch.nn.functional as F

class GradCAM:
    def __init__(self, model, target_layer):
        self.model = model
        self.activations = None

        # Récupère les activations de la couche cible
        self.handle = target_layer.register_forward_hook(
            self.save_activation
        )

    def save_activation(self, module, input, output):
        self.activations = output

    def generate(self, input_tensor, class_idx=None):
        self.model.eval()
        self.model.zero_grad()

        # Forward pass
        output = self.model(input_tensor)

        # Classe à expliquer : prédite par défaut
        if class_idx is None:
            class_idx = output.argmax(dim=1).item()

        score = output[:, class_idx]

        # Gradient du score par rapport aux activations
        gradients = torch.autograd.grad(
            outputs=score,
            inputs=self.activations
        )[0]

        # Moyenne spatiale des gradients par canal
        weights = gradients.mean(dim=(2, 3), keepdim=True)

        # Combinaison pondérée des cartes d'activation
        cam = (weights * self.activations).sum(dim=1)
        cam = F.relu(cam)

        # Normalisation entre 0 et 1
        cam = cam - cam.min()
        cam = cam / (cam.max() + 1e-8)

        # Redimensionnement à la taille de l'image
        cam = F.interpolate(
            cam.unsqueeze(1),
            size=input_tensor.shape[2:],
            mode="bilinear",
            align_corners=False
        )

        return cam.squeeze().detach().cpu().numpy(), output

    def remove_hook(self):
        self.handle.remove()
```

```python
gradcam = GradCAM(
    model=model,
    target_layer=model.layer4[-1]
)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Choisir une image du jeu de test
idx = 0

image_tensor, true_label = test_dataset[idx]

# Ajouter la dimension batch
input_tensor = image_tensor.unsqueeze(0).to(device)

# Générer Grad-CAM
cam, output = gradcam.generate(input_tensor)

# Récupérer la prédiction
predicted_label = output.argmax(dim=1).item()

# Dénormaliser l'image pour l'affichage
mean = torch.tensor([0.485, 0.456, 0.406]).view(3, 1, 1)
std = torch.tensor([0.229, 0.224, 0.225]).view(3, 1, 1)

original_image = image_tensor.cpu() * std + mean
original_image = original_image.permute(1, 2, 0).numpy()
original_image = np.clip(original_image, 0, 1)

# Affichage
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

axes[0].imshow(original_image)
axes[0].set_title("Image originale")

axes[1].imshow(cam, cmap="jet")
axes[1].set_title("Carte Grad-CAM")

axes[2].imshow(original_image)
axes[2].imshow(cam, cmap="jet", alpha=0.4)
axes[2].set_title("Superposition Grad-CAM")

for ax in axes:
    ax.axis("off")

plt.suptitle(
    f"Vérité terrain : {test_dataset.classes[true_label]} | "
    f"Prédiction : {test_dataset.classes[predicted_label]}"
)

plt.tight_layout()
plt.show()
```
